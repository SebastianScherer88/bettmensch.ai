```python
from bettmensch_ai.pipelines import OutputArtifact, InputArtifact, InputParameter

from gpt_1 import get_tokenized_data_split_and_tokenizer, pretrain_gpt_1
```

```python
get_tokenized_data_split_and_tokenizer(
    data_split="train",
    n_observations=500000,
    tokenized_data_out=OutputArtifact(name="tokenized_train_data"),
    tokenizer_out=OutputArtifact(name="tokenizer_train")
)
```

```python
get_tokenized_data_split_and_tokenizer(
    data_split="validation",
    n_observations=500000,
    tokenized_data_out=OutputArtifact(name="tokenized_validation_data"),
    tokenizer_out=OutputArtifact(name="tokenizer_validation")
)
```

```python
pretrain_gpt_1(
    tokenized_train=InputArtifact(name="tokenized_train_data"),
    tokenized_validation=InputArtifact(name="tokenized_validation_data"),
    tokenizer=InputArtifact(name="tokenizer_train"),
    dim_embed=InputParameter(name="dim_embed",value=28),
    n_decoder_layers=InputParameter(name="n_decorder_layers",value=4),
    n_heads=InputParameter(name="n_heads",value=4),
    dropout=InputParameter(name="dropout",value=0.1),
    n_epochs=InputParameter(name="n_epochs",value=10),
    batch_size=InputParameter(name="batch_size",value=2),
    shuffle=InputParameter(name="shuffle",value=True),
    display_step=InputParameter(name="display_step",value=50),
    verbose=InputParameter(name="verbose",value=False),
)
```
